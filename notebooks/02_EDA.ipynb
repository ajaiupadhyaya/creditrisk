# Credit Risk Analysis: Phase 2 - Exploratory Data Analysis (EDA)

**Objective**: Generate publication-quality visualizations of portfolio overview, credit ratings, vintage analysis, and macro stress scenarios.

**Outputs**: 10 interactive and static charts saved to `/outputs/eda/`

---

## 1. Import Libraries & Initialize

```python
import pandas as pd
import numpy as np
from pathlib import Path
import sys
import warnings
warnings.filterwarnings('ignore')

# Configure Jupyter
%matplotlib inline
%load_ext autoreload
%autoreload 2

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from audit import DataAudit
from eda import CreditRiskEDA

print("✅ All libraries imported")
```

## 2. Load Data

```python
# Load all datasets
data_path = Path.cwd().parent
audit = DataAudit(data_path)
datasets = audit.load_all_csvs()

print(f"\n📊 Loaded {len(datasets)} datasets:")
for name, df in datasets.items():
    print(f"  • {name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

## 3. Initialize EDA Module

```python
# Create output directory
output_path = data_path / 'outputs' / 'eda'
output_path.mkdir(parents=True, exist_ok=True)

# Initialize EDA
eda = CreditRiskEDA(datasets, output_path)
print(f"✅ EDA initialized")
print(f"📁 Output directory: {output_path}")
```

## 4. Portfolio Overview Visualizations

```python
print("\n📊 PORTFOLIO OVERVIEW")
print("─" * 70)

# Sunburst
fig_sunburst = eda.portfolio_sunburst()
fig_sunburst.show()
```

```python
# Loan amount distribution
fig_dist = eda.loan_amount_distribution()
fig_dist.show()
```

```python
# Correlation heatmap
eda.correlation_heatmap()
import matplotlib.pyplot as plt
plt.show()
```

```python
# Default rate by sector
fig_default = eda.default_rate_by_sector()
fig_default.show()
```

## 5. Credit Ratings Visualizations

```python
print("\n📊 CREDIT RATINGS")
print("─" * 70)

# Sankey diagram
fig_sankey = eda.rating_transition_sankey()
if fig_sankey:
    fig_sankey.show()
```

```python
# Interest rate by rating
eda.interest_rate_by_rating()
plt.show()
```

## 6. Time/Vintage Visualizations

```python
print("\n📊 TIME/VINTAGE ANALYSIS")
print("─" * 70)

# Vintage curves
fig_vintage = eda.vintage_curves()
if fig_vintage:
    fig_vintage.show()
```

```python
# Vintage heatmap
eda.vintage_heatmap()
plt.show()
```

## 7. Macro Stress Visualizations

```python
print("\n📊 MACRO STRESS SCENARIOS")
print("─" * 70)

# Macro scenarios
fig_macro = eda.macro_scenarios_detail()
if fig_macro:
    fig_macro.show()
```
