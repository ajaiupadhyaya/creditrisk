## Cell 1: Imports & Setup

```python
import sys
import pandas as pd
import numpy as np
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Add src to path
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from dashboard import DashboardAssembler

print("\n" + "="*70)
print("PHASE 6: DASHBOARD ASSEMBLY")
print("="*70 + "\n")
```

## Cell 2: Load Data

```python
# Load loan portfolio data
data_path = Path.cwd().parent / 'loan_portfolio.csv'
loan_df = pd.read_csv(data_path)

print(f"✓ Loaded {len(loan_df):,} loans from loan_portfolio.csv")
print(f"  Columns: {list(loan_df.columns)[:5]}... ({len(loan_df.columns)} total)")
```

## Cell 3: Compute Portfolio KPIs

```python
output_path = Path.cwd().parent / 'outputs'
assembler = DashboardAssembler(loan_df, output_path)

# Get KPI summary
kpis = assembler.get_portfolio_kpis()

print("\n📊 Portfolio Summary KPIs:")
print("-" * 50)
for label, value in kpis.items():
    if isinstance(value, float):
        if 'Rate' in label or 'PD' in label:
            print(f"  {label:.<40} {value:>8.2f}%")
        elif 'Exposure' in label:
            print(f"  {label:.<40} ${value:>8.1f}M")
        else:
            print(f"  {label:.<40} {value:>8.2f}")
    else:
        print(f"  {label:.<40} {value:>8,}")
print("-" * 50)
```

## Cell 4: Verify Output Files Exist

```python
# Check for outputs from previous phases
eda_files = list((output_path / 'eda').glob('*.html')) if (output_path / 'eda').exists() else []
model_files = list((output_path / 'models').glob('*.html')) if (output_path / 'models').exists() else []
stress_files = list((output_path / 'stress').glob('*.html')) if (output_path / 'stress').exists() else []
processed_files = list((output_path / 'processed').glob('*.csv')) if (output_path / 'processed').exists() else []

print("\n📁 Checking for outputs from Phases 1-5:")
print("-" * 50)
print(f"  EDA outputs (Phase 2): {len(eda_files)} files")
for f in eda_files[:3]:
    print(f"    ✓ {f.name}")
if len(eda_files) > 3:
    print(f"    ... and {len(eda_files)-3} more")

print(f"\n  Model outputs (Phase 4): {len(model_files)} files")
for f in model_files[:3]:
    print(f"    ✓ {f.name}")
if len(model_files) > 3:
    print(f"    ... and {len(model_files)-3} more")

print(f"\n  Stress outputs (Phase 5): {len(stress_files)} files")
for f in stress_files[:3]:
    print(f"    ✓ {f.name}")
if len(stress_files) > 3:
    print(f"    ... and {len(stress_files)-3} more")

print(f"\n  Processed data (Phase 3): {len(processed_files)} files")
for f in processed_files[:2]:
    print(f"    ✓ {f.name}")
print("-" * 50)
```

## Cell 5: Assemble Dashboard

```python
# Create dashboard
dashboard_file = assembler.assemble_dashboard(
    eda_path=output_path / 'eda',
    models_path=output_path / 'models',
    stress_path=output_path / 'stress'
)

print(f"\n✅ Dashboard assembly complete!")
print(f"\n   Location: {dashboard_file}")
print(f"   Size: {dashboard_file.stat().st_size / 1024:.1f} KB")
```

## Cell 6: Summary & Next Steps

```python
print("\n" + "="*70)
print("🎉 PHASE 6 SUMMARY: DASHBOARD ASSEMBLY COMPLETE")
print("="*70)

print("\n📊 Dashboard Features:")
print("   ✓ Portfolio KPI cards (6 metrics)")
print("   ✓ Interactive links to Phase 2 EDA visualizations")
print("   ✓ Model performance dashboard (Phase 4)")
print("   ✓ Stress testing results (Phase 5)")
print("   ✓ Dark theme styling (Bloomberg terminal aesthetic)")
print("   ✓ Responsive design for mobile & desktop")
print("   ✓ Fully self-contained (no CDN dependencies)")

print("\n📂 Output Location:")
print(f"   {dashboard_file}")

print("\n🚀 How to Use:")
print(f"   1. Open: {dashboard_file}")
print("   2. View portfolio KPIs")
print("   3. Click links to explore detailed analyses")
print("   4. All charts are interactive (hover, zoom, pan)")

print("\n" + "="*70)
print("✅ COMPLETE: All 6 Phases Finished Successfully")
print("="*70 + "\n")
```
