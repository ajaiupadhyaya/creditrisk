## 1. Import Libraries & Setup

```python
import pandas as pd
import numpy as np
from pathlib import Path
import sys
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
sys.path.insert(0, str(Path.cwd().parent / 'src'))

from model_training import CreditRiskModelTrainer

print("✅ Libraries imported")
```

## 2. Load Processed Data

```python
# Load processed dataset from Phase 3
data_path = Path.cwd().parent
processed_file = data_path / 'outputs' / 'processed' / 'processed_loans.csv'

df = pd.read_csv(processed_file)

print(f"✅ Processed data loaded")
print(f"📊 Shape: {df.shape}")
print(f"\nTarget variable distribution:")
print(df['defaulted'].value_counts())
print(f"\nDefault rate: {df['defaulted'].mean()*100:.2f}%")
```

## 3. Prepare Features & Target

```python
# Separate features and target
X = df.drop(columns=['defaulted', 'loan_id'], errors='ignore')
y = df['defaulted']

# Convert categorical to numeric where needed
X = X.select_dtypes(include=[np.number])

print(f"✅ Features prepared")
print(f"📊 X shape: {X.shape}")
print(f"   y shape: {y.shape}")
print(f"\n🔢 Numeric features: {X.shape[1]}")
```

```python
# Create output directory
output_path = data_path / 'outputs' / 'models'
output_path.mkdir(parents=True, exist_ok=True)

# Initialize trainer
trainer = CreditRiskModelTrainer(X, y, output_path)
print(f"✅ Model trainer initialized")
print(f"📁 Output directory: {output_path}")
```
